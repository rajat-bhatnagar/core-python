0.1 + 0.1 + 0.1 == 0.3 ## This is false ,

The expression 0.1 + 0.1 + 0.1 == 0.3 evaluates to False in Python because floating-point numbers like 0.1 cannot be represented exactly in binary inside the computer. Instead, they are stored as approximations, so the sum of three 0.1s is a number very close to but not exactly 0.3. When Python compares the sum to 0.3, it finds they are not exactly equal and returns False.

This happens because Python uses binary floating-point arithmetic (IEEE 754 standard), where decimal fractions such as 0.1 have no exact binary representation, similar to how 1/3 cannot be exactly represented in decimal. For example, 0.1 might be stored as 0.100000000000000005 and 0.3 as 0.299999999999999988 internally, so the equality check fails accurately.
